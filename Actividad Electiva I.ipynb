```python
electiva = electiva.drop_duplicates(
    subset=["id_transaccion"],
    keep="first"
)
```

```python
electiva = electiva.drop_duplicates(
    subset=["id_transaccion"]
)
```

```python
electiva.shape
```

```python
electiva["metodo_pago"] = electiva["metodo_pago"].replace({
    "Tarjeta Credito": "Tarjeta de Crédito"
})
```

```python
electiva["metodo_pago"].value_counts()
```

```python
electiva["cantidad"].median()
```

```python
electiva["cantidad"] = electiva["cantidad"].fillna(
    electiva["cantidad"].median()
)
```

```python
electiva["satisfaccion_cliente"].mode()[0]
```

```python
electiva["satisfaccion_cliente"] = (
    electiva["satisfaccion_cliente"]
    .fillna(electiva["satisfaccion_cliente"].mode()[0])
)
```

```python
electiva[electiva["precio_unitario"] == 9999.00]
```

```python
promedio_ropa = electiva[(electiva["categoria_producto"] == "Ropa") & (electiva["precio_unitario"] != 9999)]["precio_unitario"].mean()

promedio_ropa
```

```python
electiva.loc[electiva["precio_unitario"] == 9999,
    "precio_unitario"
] = promedio_ropa
```

```python
electiva["monto_total"] = (
    electiva["cantidad"] * electiva["precio_unitario"]
)
```

```python
electiva["fecha"] = pd.to_datetime(
    electiva["fecha"]
)
```

```python
electiva["dia_semana"] = (
    electiva["fecha"].dt.day_name()
)
```

```python
electiva[["fecha", "dia_semana"]].head()
```

```python
bins = [
    float("-inf"),
    99.999999,
    500,
    float("inf")
]

labels = [
    "Bajo",
    "Medio",
    "Alto"
]

electiva["rango_monto"] = pd.cut(
    electiva["monto_total"],
    bins=bins,
    labels=labels
)
```

```python
electiva["rango_monto"].value_counts()
```

```python
resumen_categoria = (
    electiva.groupby("categoria_producto")
    .agg({
        "monto_total": "sum",
        "cantidad": "mean"
    })
)

resumen_categoria
```

```python
tabla_dinamica = pd.pivot_table(
    electiva,
    values="monto_total",
    index="region",
    columns="categoria_producto",
    aggfunc="sum"
)

tabla_dinamica
```

```python
satisfaccion = (
    electiva.groupby("metodo_pago")
    ["satisfaccion_cliente"]
    .agg("mean")
)

satisfaccion
```

```python
ventas_norte = electiva[
    (electiva["region"] == "Norte") &
    (electiva["satisfaccion_cliente"] >= 4)
]

ventas_norte
```
